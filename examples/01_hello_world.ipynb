# Hello World

The smallest useful project plan: two tasks, one dependency, one schedule.

Start by importing the project planner module.

```
import project_planner as pp
from datetime import date
```

Open a workspace (in-memory for this example) and create a project starting on Monday, October 5, 2026.

```
ws = pp.open_workspace(":memory:")
ws.new_project("Hello world", start=date(2026, 10, 5))
```

Add two tasks: one to say hello (1 day), and one to celebrate (1 day).

```
say_hello = ws.add_task("Say hello", duration="1d")
celebrate = ws.add_task("Celebrate", duration="1d")
```

Link the two tasks with a Finish-to-Start dependency: Celebrate cannot start until Say hello finishes.

```
ws.add_dependency(say_hello, celebrate, "FS")
```

Run the schedule and display the result.

```
result = ws.schedule()
result
```

Print the start and finish times for each task, and the project finish.

```
project = ws.project()
node_map = {node.id: node.name for node in project.nodes}

for task in result.tasks():
    node_result = result.node(task.node_id)
    task_name = node_map[task.node_id]
    print(f"{task_name}: {node_result.start:%Y-%m-%d %H:%M} -> {node_result.finish:%Y-%m-%d %H:%M}")
print(f"Project finish: {result.project_finish:%Y-%m-%d %H:%M}")
```

Now change the first task to take 2 days instead of 1, and check the schedule again.

```
ws.set_sizing(say_hello, duration="2d")
state = ws.state()
print(f"Out of date after edit: {state.stale_dates}")
```

Reschedule with the new task duration.

```
result = ws.schedule()
print(f"New project finish: {result.project_finish:%Y-%m-%d %H:%M}")
```
